# День 2. Условия и циклы

```python
#Задание 1. Напиши программу: спроси у пользователя число через input() (не забудь про int()!), и выведи, чётное оно или нечётное. Подсказка для логики: используй оператор % (остаток от деления) — погугли, как он работает, если не знаешь.
count = int(input())
if count % 2 == 0:
  print("Четное")
else:
  print("Нечетное")
```

```python
#Задание 2. С помощью for и range() выведи все числа от 1 до 20, но пропусти (не выводи) числа, кратные 3. Понадобится и if, и, скорее всего, continue.
for i in range(1, 20):
  if i % 3 == 0:
    continue
  print(i)
```

```python
#Задание 3 (чуть сложнее). Напиши while-цикл, который спрашивает у пользователя число, и продолжает спрашивать, пока пользователь не введёт 0. Как только введён 0 — цикл останавливается и печатает "Стоп". Здесь тебе нужно будет подумать, где заводить переменную для первого запроса, чтобы цикл вообще мог начаться и проверить условие.
count = int(input())
while count != 0:
  print(count)
  count = int(input())
print("Стоп")
```

```python
#Выполняй -15 баланса пока баланс больше 0 выведи мне каждый шаг с надписью "Баланс: цифра баланса"
balance = 100
while balance > 0:
    print("Баланс:", balance)
    balance -= 15
```

# День 3. Функции

```python
#Задание 1. Напиши функцию is_even(number), которая принимает число и возвращает (через return, не print!) True, если число чётное, и False, если нечётное. Используй то, что вы уже знаете про % из дня 2. Проверь функцию, вызвав её на 2–3 разных числах и выведя результат через print(is_even(...)).
def is_even(number):
  if number % 2 == 0:
    return True
  else:
    return False

print(is_even(5))
```

```python
#Задание 2. Напиши функцию greet_person(name, age) с двумя параметрами, которая печатает фразу вида "Привет, Елена! Тебе 24 года.". Вызови её с разными именами/возрастами.
def greet_person(name = "Елена", age = "24"):
  print("Привет,", name, "!", "Тебе", age, "года.")

greet_person()
greet_person("Алла", "49")
```

```python
#Задание 3 (чуть сложнее). Напиши функцию count_vowels(word), которая принимает строку и возвращает количество гласных букв в ней (а, е, ё, и, о, у, ы, э, ю, я). Подсказка: тебе понадобится цикл for по буквам строки (как в дне 2) и счётчик-переменная, которую нужно завести внутри функции, до цикла, с начальным значением 0. Проверь на слове "программирование".
def count_vowels(word):
  vowel_count = 0
  for letter in word:
    if letter == "а" or letter == "е" or letter == "о" or letter == "и" or letter == "ё" or letter == "у" or letter == "ы" or letter == "э" or letter == "ю" or letter == "я":
      vowel_count = vowel_count + 1
  return vowel_count

print(count_vowels("яблоко"))
```
